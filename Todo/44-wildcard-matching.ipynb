# Wildcard Matching

- Pattern-matching problem with flexible symbols and full-string validation.
- Important behaviors: empty cases, runs of wildcards, and consuming zero or many characters.
- Similar ideas show up in glob filters, routing rules, and file selection logic.
- Precision matters because local wildcard choices affect the global match.
