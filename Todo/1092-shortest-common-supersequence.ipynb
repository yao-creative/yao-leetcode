# Shortest Common Supersequence

- Sequence construction problem combining two ordered inputs into one minimal result.
- The output must preserve both source orders while avoiding unnecessary duplication.
- Similar patterns appear in merge views, reconciliation tools, and change aggregation.
- This exercise is about balancing alignment quality against output length.
